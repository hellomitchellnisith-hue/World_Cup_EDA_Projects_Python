# Convert dates and seasons

```python
df["date"] = pd.to_datetime(
    df["date"],
    format="%d-%b-%y",
    errors="coerce"
)

df["season"] = df["season"].astype("string")

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month_name()

df[["date", "year", "month", "season"]].head()
```

# Extract numeric scores

```python
for col in ["team_1_score", "team_2_score"]:
    df[col + "_runs"] = pd.to_numeric(
        df[col].str.extract(r"^\s*(\d+)")[0],
        errors="coerce"
    )

df[[
    "team_1_score", "team_1_score_runs",
    "team_2_score", "team_2_score_runs"
]].head(454)
```

# DATA VISUALIZATION:
### 1.Bar plot — Top 10 teams by match wins

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
# Count recorded wins by team
wins = df["winner"].dropna().value_counts().head(10)

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    x=wins.values,
    y=wins.index,
    hue=wins.index,
    palette=sns.color_palette("rocket", n_colors=len(wins)),
    legend=False
)

# Show actual values on bars
for container in ax.containers:
    ax.bar_label(container, padding=5, fontsize=10)

plt.title("Top 10 Teams by World Cup Match Wins",
          fontsize=16, fontweight="bold")
plt.xlabel("Number of Wins")
plt.ylabel("Team")
plt.tight_layout()
plt.show()
```

# Insights: Bar Plot — Top 10 Teams by Match Wins 🏏
1. Top-performing team: The team with the tallest bar has the highest number of recorded match wins among the teams shown.

2. Performance comparison: Comparing bar lengths helps identify which teams have more wins and which have fewer wins.

3. Top 10 ranking: The chart ranks the 10 teams with the highest recorded win counts, making performance differences easy to understand.

4. Data interpretation: The analysis counts match wins, not the number of World Cup tournaments won.

5. Missing results: Matches without a recorded winner are excluded from the win counts.

* Conclusion:The bar plot compares the top 10 teams based on their recorded Cricket World Cup match wins. It helps identify the highest-performing teams, compare their winning records, and understand differences in match performance. Teams with higher bars have more recorded victories, while teams with lower bars have fewer victories among the top 10. This analysis provides a clear overview of team performance using Python, Pandas, and Seaborn.   

# 2. Histogram — Distribution of cricket scores

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
scores = df[
    ["team_1_score_runs", "team_2_score_runs"]
].melt(
    var_name="Team",
    value_name="Runs"
).dropna()

plt.figure(figsize=(12, 6))

ax = sns.histplot(
    data=scores,
    x="Runs",
    bins=20,
    color="#3478C8",
    edgecolor="white"
)

# Apply a gradient to the Seaborn histogram bars
cmap = plt.get_cmap("mako")
patches = ax.patches
heights = [p.get_height() for p in patches]
max_height = max(heights) if heights else 1

for patch in patches:
    patch.set_facecolor(
        cmap(0.25 + 0.7 * patch.get_height() / max_height)
    )

plt.title("Distribution of Cricket World Cup Team Scores")
plt.xlabel("Runs Scored")
plt.ylabel("Number of Team Scores")
plt.tight_layout()
plt.show()



```

# Insights :Histogram — Distribution of Cricket World Cup Scores
- Team 1: 445 valid scores, with an average of 238.22 runs and a median of 241 runs.
- Team 2: 440 valid scores, with an average of 195.99 runs and a median of 199.5 runs.
- Team 1's scores are higher on average by approximately 42.24 runs. This compares match positions, not specific countries.
- Team 1's highest recorded score is 417 runs, while Team 2's highest is 338 runs.
- Both groups have scores spread across a wide range, with relatively fewer observations at the very low and very high ends.
- In the combined histogram, the most frequent bin is approximately 209–230 runs, containing 115 score observations.
- Out of 885 valid team-score observations, 424 scores fall between 200 and 299 runs, making this the largest broad score category.

# 3. Scatter plot — Team 1 runs vs Team 2 runs

```python

scatter_df = df[
    ["team_1_score_runs", "team_2_score_runs"]
].dropna().copy()

scatter_df["Average Score"] = (
    scatter_df["team_1_score_runs"] +
    scatter_df["team_2_score_runs"]
) / 2

plt.figure(figsize=(11, 7))

ax = sns.scatterplot(
    data=scatter_df,
    x="team_1_score_runs",
    y="team_2_score_runs",
    hue="Average Score",
    palette="viridis",
    size="Average Score",
    sizes=(35, 150),
    alpha=0.8,
    legend="brief"
)

plt.title("Relationship Between Team 1 and Team 2 Scores")
plt.xlabel("Team 1 Runs")
plt.ylabel("Team 2 Runs")
plt.tight_layout()
plt.show()



```

# Insights: Scatter Plot — Team 1 Runs vs Team 2 Runs 🏏
- Relationship between scores: The scatter plot compares Team 1's runs with Team 2's runs in the same match.
  
- Positive correlation: If the points generally move upward from left to right, higher scores by Team 1 tend to occur alongside higher scores by Team Confirm this using the calculated correlation coefficient.

- Score variation: Widely scattered points indicate that the two teams' scores vary considerably from match to match.

- Outliers: Points far away from the main cluster represent matches with unusually high or low scores.

- Competitive matches: Points near the diagonal line \(y=x\) indicate matches where both teams scored similar numbers of runs.


# 4. Pie Chart — Cricket World Cup Match Wins by Team 🏏

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)

import matplotlib.pyplot as plt
import seaborn as sns

# Count match wins by team
wins = df["winner"].dropna().astype(str).str.strip()

# Exclude no-result entries if present
wins = wins[~wins.str.lower().isin(
    ["no result", "no winner", "abandoned", "tied"]
)]

win_counts = wins.value_counts()

# Show top 8 teams for readability
win_counts = win_counts.head(8)

# Gradient colors using Seaborn
colors = sns.color_palette(
    "mako", n_colors=len(win_counts)
)

# Create pie chart
plt.figure(figsize=(10, 8))

plt.pie(
    win_counts.values,
    labels=win_counts.index,
    colors=colors,
    autopct="%1.1f%%",
    startangle=140,
    pctdistance=0.70,
    wedgeprops={"edgecolor": "white", "linewidth": 2},
    textprops={"fontsize": 10}
)

plt.title(
    "Top 8 Teams by Recorded World Cup Match Wins",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()
plt.show()


```

# Insights: Pie Chart — Cricket World Cup Match Wins 🏏
- Australia leads with 69 recorded wins, the largest slice of the pie chart.

- New Zealand ranks second with 54 wins, closely followed by India with 53.

- India and New Zealand have nearly equal shares, differing by just one recorded win.

- South Africa has the fewest wins among these eight teams, with 38.

- Australia has 31 more wins than South Africa in this comparison.

- The top 3 teams—Australia, New Zealand and India—account for 45.1% of the combined wins of the selected top 8 teams.
